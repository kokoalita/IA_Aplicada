```python
from dotenv import load_dotenv
load_dotenv()
```

```python
import anthropic

client = anthropic.Anthropic()  # lee ANTHROPIC_API_KEY

msg = client.messages.create(
    model="claude-opus-5-5",
    max_tokens=1000,
    messages=[{"role": "user", "content": "Explain in two sentences what a token is."}],
)
print(msg.content[0].text)
print(msg.usage)        # input_tokens, output_tokens: lo que pagas
print(msg.stop_reason)  # end_turn si terminó bien
```
